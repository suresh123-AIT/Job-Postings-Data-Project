```python
#import libraries 
import os 
import pandas as pd 
from datasets import load_dataset 
import matplotlib.pyplot as plt 
import ast
import seaborn as sns

#path for local caching 

file_path="data_jobs.parquet"

#loading data 
if not os.path.exists(file_path):
    dataset=load_dataset('lukebarousse/data_jobs') 
    df=dataset['train'].to_pandas()
    df.to_parquet(file_path)
else:
    df=pd.read_parquet(file_path)

#data cleanup 
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda skill_list2: ast.literal_eval(skill_list2) if isinstance(skill_list2,str) else skill_list2 )


```

```python
df_DA_US=df[(df['job_title_short']=='Data Analyst')&(df['job_country']=='United States')].copy()
```

```python
df_DA_US['job_posted_month_no']=df_DA_US['job_posted_date'].dt.month
```

```python
df_DA_US_Explode=df_DA_US.explode('job_skills')
```

```python
df_DA_US_pivot=df_DA_US_Explode.pivot_table(index='job_posted_month_no',columns='job_skills',aggfunc='size',fill_value=0)

df_DA_US_pivot.loc['Total']=df_DA_US_pivot.sum()

df_DA_US_pivot=df_DA_US_pivot[df_DA_US_pivot.loc['Total'].sort_values(ascending=False).index]
df_DA_US_pivot=df_DA_US_pivot.drop('Total')
df_DA_US_pivot
```

```python
df_DA_US_pivot.plot(kind='line')
```

```python
DA_Totals=df_DA_US.groupby('job_posted_month_no').size()
```

```python
df_DA_US_Percent=df_DA_US_pivot.div(DA_Totals/100,axis=0)
df_DA_US_Percent
```

```python
df_DA_US_Percent=df_DA_US_Percent.reset_index()
df_DA_US_Percent['job_posted_month']=df_DA_US_Percent['job_posted_month_no'].apply(lambda x:pd.to_datetime(x,format='%m').strftime('%b'))
df_DA_US_Percent=df_DA_US_Percent.set_index('job_posted_month')
df_DA_US_Percent=df_DA_US_Percent.drop(columns='job_posted_month_no')
df_DA_US_Percent
```

```python
df_plot=df_DA_US_Percent.iloc[:,:5]
sns.lineplot(data=df_plot,dashes=False,palette='tab10')
sns.set_theme(style='ticks')
sns.despine()

plt.title('Trending Top Skills for Data Analysts in the US')
plt.ylabel('Likelyhood in Job Posting')
plt.xlabel('2023')
plt.legend().remove()

from matplotlib.ticker import PercentFormatter
ax=plt.gca()
ax.yaxis.set_major_formatter(PercentFormatter(decimals=0))

# for i in range(5):
#     plt.text(11.2,df_plot.iloc[-1,i],df_plot.columns[i]) 
lines=ax.get_lines() 
offsets={
    'python':-0.5,
    'tableau':1.0
}

for i in range(5):
    col_name=df_plot.columns[i]
    y_val=df_plot.iloc[-1,i]+offsets.get(col_name,0.0)
    plt.text(
        11.2,
        y_val,
        col_name,
        color=lines[i].get_color(),
        va='center',
        fontsize=10,
        fontweight='bold'
    )
```
